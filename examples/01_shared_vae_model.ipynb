# VAE model example

Build a compact VAE and inspect its reconstruction and latent outputs.

```python
import torch
from nuclear_vae_embeddings.models import VAE
model = VAE(nc=1, ngf=8, ndf=8, latent_variable_size=16, imsize=64, batchnorm=False)
model.eval()
```

```python
with torch.no_grad():
    reconstruction, latent, mean, log_variance = model(torch.zeros(2, 1, 64, 64))
reconstruction.shape, latent.shape
```
