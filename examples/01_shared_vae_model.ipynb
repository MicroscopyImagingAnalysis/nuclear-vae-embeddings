# VAE model example

Fit a compact VAE, reconstruct a batch and extract identity-preserving latent features.

```python
import torch
from torch.utils.data import DataLoader, Dataset
from nuclear_vae_embeddings import fit_vae, latent_feature_table, reconstruct_images
from nuclear_vae_embeddings.models import VAE

class ExampleCrops(Dataset):
    def __len__(self):
        return 8

    def __getitem__(self, index):
        image = torch.zeros(1, 64, 64)
        image[:, 20:44, 20:44] = 0.4 + 0.05 * index
        return {'image': image, 'object_id': f'nucleus-{index}', 'condition': index // 4}

loader = DataLoader(ExampleCrops(), batch_size=4, shuffle=False)
model = VAE(nc=1, ngf=8, ndf=8, latent_variable_size=16, imsize=64, batchnorm=False)
```

```python
optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)
history = fit_vae(model, loader, optimizer, epochs=2)
history
```

```python
example_images = next(iter(loader))['image']
reconstruction = reconstruct_images(model, example_images)
reconstruction.shape
```

```python
latent_table = latent_feature_table(model, loader, metadata_keys=['condition'])
latent_table.head()
```
